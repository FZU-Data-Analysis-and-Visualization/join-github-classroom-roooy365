```python
#一 1
# 导入pandas库
import pandas as pd

# 定义数据字典
data_dict = {
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
}

# 用字典创建DataFrame
df1 = pd.DataFrame(data_dict)

# 查看创建结果
print("df1 数据：")
print(df1)
```

```python
#一 2
# 查看前1行
print("前1行数据：")
print(df1.head(1))

# 查看DataFrame的形状（行数, 列数）
print("\nDataFrame形状：", df1.shape)

# 查看列名
print("\n列名：", df1.columns.tolist())

# 查看每列的数据类型
print("\n数据类型：")
print(df1.dtypes)
```

```python
#一 3
# 定义列表数据
data_list = [[1, "A"], [2, "B"], [3, "C"]]

# 创建DataFrame并指定列名
df2 = pd.DataFrame(data_list, columns=["id", "tag"])

# 查看创建结果
print("df2 数据：")
print(df2)
```

```python
#二 1
# 方法1：直接列名选取
df_col = df1[["score", "gender"]]
print("选取score和gender列：")
print(df_col)

# 方法2：用loc（推荐，更灵活）
df_col2 = df1.loc[:, ["score", "gender"]]  #:表示选取所有行
print("\n用loc选取的结果：")
print(df_col2)
```

```python
#二 2
# 方法1：用iloc按行索引选取
df_row = df1.iloc[[0, 2]]
print("选取索引0、2的行：")
print(df_row)

# 方法2：用loc（支持任意类型的标签（字符串、时间、自定义值等））
df_row2 = df1.loc[[0, 2]]
print("\n用loc选取的结果：")
print(df_row2)
```

```python
#二 3
# 布尔索引筛选
df_filter = df1[df1["score"] > 80]
print("筛选score>80的行：")
print(df_filter)
```

```python
#二 4
# 方法1：用apply函数
df1["grade"] = df1["score"].apply(lambda x: "优秀" if x >= 90 else "良好")

# 方法2：用numpy.where（更高效）
# import numpy as np
# df1["grade"] = np.where(df1["score"] >= 90, "优秀", "良好")

print("新增grade列后的df1：")
print(df1)
```

```python
#二 5
# 方法1：用drop函数（axis=1表示删除列，inplace=True表示直接修改原DataFrame）
df1.drop("gender", axis=1, inplace=True)

# 方法2：重新赋值（不修改原数据，更安全）
# df1 = df1.drop("gender", axis=1)

print("删除gender列后的df1：")
print(df1)
```

```python
#三 1
import pandas as pd

# 1. 创建含缺失值的DataFrame
data_missing = {"x": [1, None, 3], "y": [None, 5, 6]}
df_missing = pd.DataFrame(data_missing)
print("=== 含缺失值的DataFrame ===")
print(df_missing)

# 检测缺失值（返回布尔矩阵，True表示该位置为缺失值）
print("\n=== 缺失值检测结果（True代表缺失） ===")
print(df_missing.isnull())

# 统计每列缺失值的数量
print("\n=== 每列缺失值数量统计 ===")
print(df_missing.isnull().sum())
```

```python
#三 2
# 用0填充所有缺失值（会生成新的DataFrame，不修改原数据）
df_fill = df_missing.fillna(0)
print("=== 用0填充缺失值后的DataFrame ===")
print(df_fill)

# 删除所有包含缺失值的行（同样生成新的DataFrame）
df_drop = df_missing.dropna()
print("\n=== 删除含缺失值的行后的DataFrame ===")
print(df_drop)
```

```python
#四 1
# 还原原始DataFrame（name, score, gender）
df1 = pd.DataFrame({
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
})
# 按gender分组，同时计算score的均值和最大值
group_result = df1.groupby("gender")["score"].agg(["mean", "max"])
print("=== 按gender分组的score均值和最大值 ===")
print(group_result)
```

```python
#四 2
# 方法1：用groupby + size()统计每组行数
group_count = df1.groupby("gender").size()
print("\n=== 按gender分组的每组人数 ===")
print(group_count)

# 方法2：直接用value_counts()（更简洁）
#print(df1["gender"].value_counts())
```

```python
#五 1
# 按score列升序排序（ascending=True为默认值，可省略）
df_sorted = df1.sort_values(by="score", ascending=True)
print("=== 按score升序排序后的DataFrame ===")
print(df_sorted)
```

```python
#五 2
# 1. 添加重复数据：Tom, 85, 男
new_row = pd.DataFrame({"name": ["Tom"], "score": [85], "gender": ["男"]})
df_duplicate = pd.concat([df1, new_row], ignore_index=True)
print("\n=== 添加重复数据后的DataFrame ===")
print(df_duplicate)

# 2. 去重（默认按所有列判断重复，保留第一次出现的行）
df_dedup = df_duplicate.drop_duplicates()
print("\n=== 去重后的DataFrame ===")
print(df_dedup)
```

```python
#六 1
import pandas as pd

# 定义两个待合并的DataFrame
df1 = pd.DataFrame({"id": [1, 2], "name": ["X", "Y"]})
df2 = pd.DataFrame({"id": [1, 2], "age": [20, 30]})

# 按id进行内连接合并（默认how='inner'，也可指定how='left'/'right'/'outer'）
merged_df = pd.merge(df1, df2, on="id")

print("=== 横向合并结果 ===")
print(merged_df)
```

```python
#六 2
# 定义两个待拼接的DataFrame
df_a = pd.DataFrame({"val": [1, 2]})
df_b = pd.DataFrame({"val": [3, 4]})

# 纵向拼接并重置索引（两种等价写法）
# 写法1：直接在concat时重置索引
concat_df = pd.concat([df_a, df_b], ignore_index=True)

# 写法2：拼接后再用reset_index重置索引
# concat_df = pd.concat([df_a, df_b])
# concat_df = concat_df.reset_index(drop=True)  # drop=True删除原索引列

print("\n=== 纵向拼接结果（已重置索引） ===")
print(concat_df)
```

```python
#七 1
# 创建时间字符串列表
time_list = ["2025-01-01", "2025-01-02", "2025-01-03"]

# 转为datetime格式（返回DatetimeIndex）
time_series = pd.to_datetime(time_list)

# 转为DataFrame方便后续处理
df_time = pd.DataFrame({"date": time_series})

print("=== 时间列（datetime格式） ===")
print(df_time)
print("\n时间列数据类型：", df_time["date"].dtype)
```

```python
#七 2
# 提取月份（数字1-12）
df_time["month"] = df_time["date"].dt.month

# 提取星期（两种方式：数字/名称）
df_time["weekday_num"] = df_time["date"].dt.weekday  # 0=周一，1=周二...6=周日
df_time["weekday_name"] = df_time["date"].dt.day_name()  # 直接获取星期名称（如Wednesday）

print("\n=== 提取月份与星期结果 ===")
print(df_time)
```

```python
#八 1
df1 = pd.DataFrame({
    "name": ["Tom", "Jerry", "Mike"],
    "score": [85, 92, 78],
    "gender": ["男", "女", "男"]
})
# 计算均值和标准差
score_mean = df1["score"].mean()
score_std = df1["score"].std()

# 标准化处理
df1["score_std"] = (df1["score"] - score_mean) / score_std

print("=== score列标准化结果 ===")
print(df1[["name", "score", "score_std"]])
```

```python
#八 2
# 创建透视表
pivot_table = df1.pivot_table(
    values="score",    # 要聚合的值列
    index="gender",    # 行索引
    aggfunc="median"   # 聚合函数：中位数
)

print("\n=== 按gender分组的score中位数透视表 ===")
print(pivot_table)
```
